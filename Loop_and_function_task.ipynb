### **Loop and Function Assessment**

```python
# 1. Print numbers from 1 to 10 using a for loop.

for i in range(1,10+1):
    print(i,end=" ")
```

```python
# 2. Print numbers from 10 to 1 using a while loop.

a = 10
while a >= 1:
    print(a,end=" ")
    a -= 1
```

```python
# 3. Write a program to print the multiplication table of a given number.

a = int(input("Enter the number: "))
for i in range(1,10+1):
    print(i, "x", a, "=",i * a)
```

```python
# 4. Write a program to find the sum of first N natural numbers using a loop.

a = int(input("Enter the number: "))
sum = 0
for i in range(1,a+1):
    sum = sum + 1
    print(sum,end=" ")
print("\nthe sum of first N natural numbers of given number is:",a)
```

```python
# 5. Print the following pattern using a loop:#

for i in range(5):
    for j in range(0,i+1):
        print("#",end="")
    print()
```

```python
# 6. Write a function greet(name) that prints "Hello, <name>"

def greet(name):
    print("Hello,",name)
greet("bratheep")
```

```python
# 7. Write a function add_numbers(a, b) that returns the sum of two numbers.

a = int(input("Enter A number:"))
b = int(input("Enter B number:"))
def add_numbers(a,b):
    return a + b
print(add_numbers(a,b))
```

```python
# 8. Write a function is_even(n) that returns True if the number is even, otherwise False.

n = int((input("Enter the number:")))
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False
print("The given number is:",n)
print(is_even(n))
```

```python
# 9. Write a function square(n) that returns the square of a given number.

n = int(input("Enter the number:"))
def square(n):
    return n ** 2
print("The given number:",n)
print("The square of the given number is:",square(n))
```

```python
# 10. Write a function max_of_two(a, b) that returns the larger of two numbers

a = int(input("Enter the A number:"))
b = int(input("Enter the B number:"))

def max_of_two(a,b):
    if a > b:
        return a
    else:
        return b
print("A =",a)
print("B =",b)
print("The larger of two numbers is:",max_of_two(a,b))
```
