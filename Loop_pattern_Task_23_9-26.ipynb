### **Loop Pattern Task**
### ***While loop pattern***

```python
a = 1
while a<=5:
    b = 1
    while b <= a:
        print("*",end='')
        b += 1
    print()
    a += 1
```

```python
x = int(input("Enter the row: "))
symbol = input("Enter the symbol")
a = 1
while a <= x:
    b = 1
    while b <= a:
        print(symbol,end='')
        b += 1
    print()
    a += 1
```

### ***For Loop pattern*** 

```python
for i in range(6):
    for j in range(0,i+1):
        print("*",end='')
    print()
```

```python
a = int(input("enter the row:"))
symbol = input("Enter the symbol:")

for i in range(a):
    for j in range(0,i+1):
        print(symbol,end='')
    print()
```
