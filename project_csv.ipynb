```python

import os
import csv
import re
import ipaddress

path = "D:\\Gen AI"
filename = "Linux_2k.log"

ip_address = []

# Read the log file
with open(os.path.join(path, filename), "r") as f:

    for each in f:

        if "sshd(pam_unix)" in each and "authentication failure" in each:

            match = re.search(r"rhost=([^\s]+)", each)

            if match:
                ip = match.group(1)

                # Check whether it is a valid IP address
                try:
                    ipaddress.ip_address(ip)
                    ip_address.append(ip)

                except ValueError:
                    continue


# Count IP address frequency
def frequency(mylist):

    mydict = {}

    for each in mylist:

        if each not in mydict:
            mydict[each] = 1
        else:
            mydict[each] += 1

    return mydict


# Convert dictionary into CSV
def to_csv(mydict):

    output_file = os.path.join(path, "failed_ssh_attempts.csv")

    with open(output_file, "w", newline="") as f:

        writer = csv.writer(f)

        writer.writerow(["S.No", "IP_Address", "Failed_Login_Count"])

        for sno, (ip, count) in enumerate(
            sorted(mydict.items(), key=lambda x: x[1], reverse=True),
            start=1
        ):

            writer.writerow([sno, ip, count])

    print("CSV file created successfully:", output_file)


# Function calling
x = frequency(ip_address)

print("Total failed SSH attempts:", len(ip_address))
print("Unique IP addresses:", len(x))

print("\nIP Address Frequency:")

for ip, count in x.items():
    print(ip, ":", count)
to_csv(x)

```
