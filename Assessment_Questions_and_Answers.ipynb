### **Python Assessment Questions Part -03**

```python
#1.Write a Python program to find the length of a list using len().

num = [20,30,40,50,60,70]
print("Length of list:", len(num))
```

```python
#2.Write a Python program to find the maximum and minimum value in a list

num = [20,30,40,50,60]
print("Maximum value in the list:", max(num))
print("Minimum value in the list:", min(num))
```

```python
#3.Use sum() to calculate the total of all numbers in a list.

num = [20,30,40,50,60]
Total = sum(num)
print("Total:",Total)
```

```python
#4.Create a list of numbers and sort them in ascending and descending order using sort().

num = [2,9,6,8,4,1,3,5,7]
num.sort()
print("Ascending order:", num)
num.sort(reverse = True)
print("Descending order:", num)
```

```python
#5.Write a Python program to create a tuple and print it.

my_tuple = (20,30,40,50,60)
print("Tuple:", my_tuple,type(my_tuple))
```

```python
#6.Write a program to access the first, middle, and last elements of a tuple using indexing.

my_tuple = (20,30,40,50,60)
print("First element:", my_tuple[0])
print("Middle element:", my_tuple[len(my_tuple) // 2])
print("Last element:", my_tuple[-1])
```

```python
#7.Write a program to slice a tuple (first 3 elements, last 3 elements).

my_tuple = (10,20,30,40,50,60)
print("First 3 elements:", my_tuple[:3])
print("Last 3 elements:", my_tuple[-3:])
```

```python
#8.Write a program to remove an element from a set using discard().

my_set = {10,20,30,40,50}
my_set.discard(20)
print("Updated set:", my_set)
```

```python
#9.Write a program to pop a random element from a set using pop().

my_set = {10,20,30,40,50}
removed = my_set.pop()

print("Removed element:", removed)
print("Updated set:", my_set)
```

```python
#10.Write a program to update the value of an existing key in a dictionary.

student = {"Name": "Pradeep",
            "Age": 22, 
            "City": "Chennai"}
student["City"] = "Sivakasi"
print(student)
```

```python
#11.Write a program to remove the last inserted key-value pair using popitem().

student = {"Name": "Pradeep",
            "Age": 22, 
            "City": "Chennai"}
student.popitem()
print(student)
```

```python
#12.Write a program to remove all key-value pairs using clear().

student = {"Name": "Pradeep",
            "Age": 22, 
            "City": "Chennai"}
student.clear()
print(student)
```

```python
#13.Write a program to get all keys using keys().

student = {"Name": "Pradeep",
            "Age": 22, 
            "City": "Chennai"}

print(student.keys())
```

```python
#14.Write a program to get all key-value pairs using items().

student = {"Name": "Pradeep",
            "Age": 22, 
            "City": "Chennai"}
print(student.items())
```

```python
#15.Write a program to check if a key exists in a dictionary using in.

student = {"Name": "Pradeep",
            "Age": 22, 
            "City": "Chennai"}
if "City" in student:
    print("Key exists")
else:
    print("Not exists")
```
