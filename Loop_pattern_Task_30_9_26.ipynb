```python
x = int(input("Enter the row:"))
symbol = input("Enetr the symbol:")
for i in range(1,x + 1):
    for j in range(1, x - i + 1):
        print(" ",end="")
    for j in range(1,i + 1):
        print(symbol,end="")
    print()
```

### ***While loop***

```python
x = int(input("Enter the row:"))
symbol = input("Enter the symbol:")
a = 1
while a <= x:
    print(" " * (x-a), end="")
    
    b = 1
    while b <= a:
        print(symbol,end="")
        b += 1
    a += 1
    print()
    
```
