### ***For Loop***

```python
n = int(input("enter the number 0-9:"))
for i in range(n+1):
    for j in range(i):
        print(i,end="")
    print()
    
```

### ***While Loop***

```python
n = int(input("enter the number 0-9:"))
a = 1
while a <= n:
    b = 1
    while b <= a:
        print(a,end="")
        b += 1
    print()
    a += 1
```
