### **For Loop**

```python
x = int(input("Enter the row: "))
symbol = input("Enter the symbol")
for i in range(x,0,-1):
    for i in range(i):
        print(symbol,end="")
    print()
```

### **While Loop**

```python
x = int(input("Enter the row: "))
symbol = input("Enter the symbol")
a = x
while a >= 1:
    b = 1
    while b <= a:
        print(symbol,end='')
        b += 1
    print()
    a -= 1
```
