### **Import datetime**

```python
import datetime
ans = datetime.datetime(1981, 1, 20)
print(ans)
ans1 = datetime.datetime.now()
print(ans1)
print(ans.strftime("%w"))
print(ans.strftime("%a"))
print(ans.strftime("%A"))
print(ans.strftime("%d"))
print(ans.strftime("%D"))
print(ans.strftime("%Y"))
```

### **Json**

```python
import json
mydict={
    "name": "pradeep",
    "age": 22,
    "degree": "B.E.E.C.E",
    "city": "Sivakasi",
    "year": None
}
with open("C:\\Users\\nmdos\\OneDrive\\Desktop\\jk_1.txt",'w') as json_file:
    json.dump(mydict, json_file)
json_string = json.dumps(mydict)
print(json_string, type(json_string))
dict_from_json = json.loads(json_string)
print(dict_from_json, type(dict_from_json))
```
