### **File Handling**

```python
myfile = open("C:/Users/nmdos/OneDrive/Desktop/jk.txt")
print(myfile.tell())
print("------------------------")
print(myfile.read(100))
print("------------------------")
print(myfile.read())
print(myfile.tell())
myfile.close()
```

### **with - keyword** 

```python
with open("C:\\Users\\nmdos\\OneDrive\\Desktop\\jk.txt") as myfile:
    read_line = myfile.readline()
    print(read_line)
```

```python
with open("C:\\Users\\nmdos\\OneDrive\\Desktop\\jk.txt") as myfile:
    read_line = myfile.readlines()
    print(read_line)
    print("------------")
    print(read_line[0])
    print("------------")
    for each in read_line:
        print(each)
```

### **"a" - Append Mode**

```python
with open("C:\\Users\\nmdos\\OneDrive\\Desktop\\jk.txt",'a') as myfile:
    mywrite = myfile.write("\nhello")
```

### **"w" - Write Mode**

```python
with open("C:\\Users\\nmdos\\OneDrive\\Desktop\\jk.txt",'w') as myfile:
    mywrite = myfile.write("Hello")
    mywrite = myfile.write("\nIts pradeep")
    mywrite = myfile.write("\ncity: Sivakasi")
```
